```python
from sagemaker.xgboost import XGBoost

role = "arn:aws:iam::666258711441:role/service-role/AmazonSageMaker-ExecutionRole-20260722T143599"

estimator = XGBoost(
    entry_point="train.py",
    framework_version="1.7-1",
    instance_type="ml.m5.large",
    instance_count=1,
    role=role,
    hyperparameters={"max_depth": 5, "eta": 0.2, "objective": "binary:logistic", "num_round": 100},
)

estimator.fit({
    "train": "s3://beant-mlops-portfolio-666258711441/processed/train/",
    "validation": "s3://beant-mlops-portfolio-666258711441/processed/test/",
})
```

```python
from sagemaker.experiments.run import Run

with Run(experiment_name="churn-xgboost-experiments", run_name="run-1") as run:
    estimator.fit({
        "train": "s3://beant-mlops-portfolio-666258711441/processed/train/",
        "validation": "s3://beant-mlops-portfolio-666258711441/processed/test/",
    })
```

```python
from sagemaker.tuner import HyperparameterTuner, ContinuousParameter, IntegerParameter

hyperparameter_ranges = {
    "eta": ContinuousParameter(0.1, 0.3),
    "max_depth": IntegerParameter(3, 7),
}

tuner = HyperparameterTuner(
    estimator=estimator,
    objective_metric_name="validation:f1",
    metric_definitions=[{"Name": "validation:f1", "Regex": r"validation:f1=([0-9\.]+)"}],
    objective_type="Maximize",
    hyperparameter_ranges=hyperparameter_ranges,
    max_jobs=4,
    max_parallel_jobs=2,
)

tuner.fit({
    "train": "s3://beant-mlops-portfolio-666258711441/processed/train/",
    "validation": "s3://beant-mlops-portfolio-666258711441/processed/test/",
})
```
